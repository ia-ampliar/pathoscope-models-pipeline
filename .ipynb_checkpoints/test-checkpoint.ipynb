### **Paths e hiperparâmetros**

```python
@dataclass(frozen=True)
class TrainConfig:
    # Paths
    root: str
    data_dir: str
    split_dir: str
    image_root: str
    runs_dir: str

    train_csv: str
    val_csv: str
    test_csv: str

    # Imagem / batch
    img_height: int = 224
    img_width: int = 224
    batch_size: int = 32

    # Treino baseline
    epochs_baseline: int = 100
    lr_baseline: float = 1e-4

    # Fine-tuning
    do_finetune: bool = True
    fine_tune_percent: float = 0.30      # ~30% final do backbone
    lr_fine: float = 1e-5
    epochs_fine: int = 30

    # Callbacks
    early_stop_patience: int = 10
    monitor: str = "val_loss"

    # Augment
    augment: bool = True

    # QAT (opcional; pode exigir legacy Keras dependendo do TFMOT)
    do_qat: bool = True
    qat_epochs: int = 15
    lr_qat: float = 1e-5
    qat_patience: int = 5


# --- Preencha/ajuste aqui conforme seu ambiente ---
ROOT = Path("/home/ampliar/Desktop/train-qat-models/train-pipeline/")
DATA_DIR = ROOT / "Datas"
SPLIT = ROOT / "split"

cfg = TrainConfig(
    root=str(ROOT),
    data_dir=str(DATA_DIR),
    split_dir=str(SPLIT),
    image_root=str(DATA_DIR),
    runs_dir=str(ROOT / "runs"),

    train_csv=str(SPLIT / "train_data.csv"),
    val_csv=str(SPLIT / "val_data.csv"),
    test_csv=str(SPLIT / "test_data.csv"),

    img_height=224,
    img_width=224,
    batch_size=32,

    epochs_baseline=100,
    lr_baseline=1e-4,

    do_finetune=True,
    fine_tune_percent=0.30,
    lr_fine=1e-5,
    epochs_fine=30,

    early_stop_patience=10,
    monitor="val_loss",
    augment=True,

    do_qat=True,
    qat_epochs=15,
    lr_qat=1e-5,
    qat_patience=5,
)

cfg
```

```python
# --- Pasta da execução (run_dir) ---
run_name = f"mobilenetv2_{time.strftime('%Y%m%d_%H%M%S')}"
run_dir = Path(cfg.runs_dir) / run_name
(run_dir / "checkpoints").mkdir(parents=True, exist_ok=True)
(run_dir / "metrics").mkdir(parents=True, exist_ok=True)

with open(run_dir / "config.json", "w", encoding="utf-8") as f:
    json.dump(asdict(cfg), f, indent=2, ensure_ascii=False)

print("Run dir:", run_dir)
```

```python
def safe_read_csv(path: str) -> pd.DataFrame:
    df = pd.read_csv(path)
    required = {"image_path", "lesion_type"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"{path} sem colunas obrigatórias: {missing}")

    df = df.copy()
    df["image_path"] = df["image_path"].astype(str).str.strip()
    df["lesion_type"] = df["lesion_type"].astype(str).str.strip()
    df = df.dropna(subset=["image_path", "lesion_type"])
    df = df[(df["image_path"] != "") & (df["lesion_type"] != "")]
    return df.reset_index(drop=True)


train_df = safe_read_csv(cfg.train_csv)
val_df   = safe_read_csv(cfg.val_csv)
test_df  = safe_read_csv(cfg.test_csv)

classes = sorted(train_df["lesion_type"].unique().tolist())
num_classes = len(classes)
class_to_idx = {c: i for i, c in enumerate(classes)}

print(f"Treino: {len(train_df)} | Val: {len(val_df)} | Teste: {len(test_df)}")
print("Classes:", classes)
print("num_classes:", num_classes)

with open(run_dir / "classes.json", "w", encoding="utf-8") as f:
    json.dump({"classes": classes, "class_to_idx": class_to_idx}, f, indent=2, ensure_ascii=False)
```

```python
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

AUTOTUNE = tf.data.AUTOTUNE
IMG_SIZE = (cfg.img_height, cfg.img_width)

def make_dataset(df: pd.DataFrame, *, image_root: str, shuffle: bool, augment: bool) -> tf.data.Dataset:
    paths = df["image_path"].values.astype(str)
    labels_int = df["lesion_type"].map(class_to_idx).values.astype(np.int32)
    labels_oh = keras.utils.to_categorical(labels_int, num_classes=num_classes).astype(np.float32)

    ds = tf.data.Dataset.from_tensor_slices((paths, labels_oh))

    if shuffle:
        ds = ds.shuffle(buffer_size=len(df), reshuffle_each_iteration=True)

    def _load(path, label):
        if image_root and image_root.strip():
            path = tf.strings.join([str(image_root).rstrip("/") + "/", path])
        img = tf.io.read_file(path)

        # Se suas imagens forem PNG, troque para decode_image.
        img = tf.image.decode_jpeg(img, channels=3)
        img = tf.image.resize(img, IMG_SIZE)
        img = tf.cast(img, tf.float32)

        # ✅ Normalização correta p/ MobileNetV2: [-1, 1]
        img = preprocess_input(img)
        return img, label

    def _augment(img, label):
        img = tf.image.random_flip_left_right(img)
        img = tf.image.random_brightness(img, 0.10)
        img = tf.image.random_contrast(img, 0.90, 1.10)
        return img, label

    ds = ds.map(_load, num_parallel_calls=AUTOTUNE)
    if augment and shuffle:
        ds = ds.map(_augment, num_parallel_calls=AUTOTUNE)

    ds = ds.batch(cfg.batch_size, drop_remainder=False).prefetch(AUTOTUNE)
    return ds
```

```python
train_ds = make_dataset(train_df, image_root=cfg.image_root, shuffle=True, augment=cfg.augment)
val_ds   = make_dataset(val_df,   image_root=cfg.image_root, shuffle=False, augment=False)
test_ds  = make_dataset(test_df,  image_root=cfg.image_root, shuffle=False, augment=False)
```
